### Problem Statement
Financial institutions face significant challenges in managing delinquent loans and optimizing their collections strategies. Inefficient identification of high-risk accounts and lack of visibility into repayment behavior often lead to increased defaults, reduced recovery rates, and financial losses.

This project aims to analyze historical loan data from a peer-to-peer lending platform to evaluate the effectiveness of current collections performance. The objective is to track borrower repayment patterns, identify stages of delinquency, and measure key metrics such as recovery rates, roll rates (movement across delinquency buckets), and charge-off trends.

By developing an interactive dashboard, this study seeks to provide actionable insights into:
- The progression of loans from current to delinquent and default states
- Segmentation of borrowers based on repayment behavior and risk exposure
- Identification of critical points where collections interventions can be most effective
- Evaluation of portfolio-level risk and recovery performance

The final outcome will support data-driven decision-making for improving collections strategies, minimizing credit losses, and enhancing overall portfolio health.

```python
df['pub_rec_bankruptcies'] = df['pub_rec_bankruptcies'].fillna(df['pub_rec_bankruptcies'].median())
```

```python
df['mort_acc'] = df['mort_acc'].fillna(df['mort_acc'].median())
```

```python
df['revol_util']= df['revol_util'].fillna(df['revol_util'].median())
```

```python
df['emp_title'] = df['emp_title'].fillna(df['emp_title'].mode().iloc[0])
df['emp_length'] = df['emp_length'].fillna(df['emp_length'].mode().iloc[0])
df['title'] = df['title'].fillna(df['title'].mode().iloc[0])
```

```python
(df.isnull().sum()/len(df))*100
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace = True)
```

```python
df['loan_status'].value_counts()
```

```python
if 'int_rate' in df.columns:
    df['int_rate'] = df['int_rate'].astype(str).str.replace('%','')
    df['int_rate'] = pd.to_numeric(df['int_rate'], errors='coerce')
```

```python
df['loan_status'] = df['loan_status'].str.strip()
```

```python
df['is_default'] = np.where(df['loan_status'] == 'Charged Off', 1, 0)
df['is_late'] = df['loan_status'].str.contains('Late', case=False).astype(int)
df['is_recovered'] = np.where(df['loan_status'] == 'Fully Paid', 1, 0)
```

```python
if 'annual_inc' in df.columns:
    df['income_group'] = pd.cut(df['annual_inc'],
    bins=[0, 50000, 100000, np.inf],
    labels=['Low', 'Medium', 'High'])
```

```python
if 'loan_amnt' in df.columns:
    df['loan_size'] = pd.cut(df['loan_amnt'],
    bins=[0, 5000, 15000, np.inf],
    labels=['Small', 'Medium', 'Large'])
```

```python
df['high_risk_flag'] = np.where(
    (df['is_default'] == 1) | (df['is_late'] == 1),
    1, 0
)
```

```python
if 'issue_d' in df.columns:
    df['issue_d'] = pd.to_datetime(df['issue_d'], errors='coerce')
    df['issue_year'] = df['issue_d'].dt.year
    df['issue_month'] = df['issue_d'].dt.month
```

```python
df.drop_duplicates(inplace = True)
```

```python
print("\nFinal Shape:", df.shape)
print("\nCleaned Data Sample:\n", df.head())
```

```python
df.to_csv("cleaned_loans.csv", index=False)
```
